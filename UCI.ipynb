### Loading the Bank Marketing Dataset

First, we'll download the 'bank-additional' dataset, which is a version of the Bank Marketing Dataset from the UCI Machine Learning Repository. This dataset contains information about direct marketing campaigns of a Portuguese banking institution. The goal is to predict whether a client will subscribe to a term deposit ('yes' or 'no').

```python
import pandas as pd
import requests
import zipfile
import io

# URL for the bank marketing dataset (bank-additional.zip)
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/00222/bank-additional.zip"

# Download the zip file
response = requests.get(url)
response.raise_for_status() # Check if the request was successful

# Unzip the file in memory
zip_file = zipfile.ZipFile(io.BytesIO(response.content))

# The dataset is usually in a CSV file named 'bank-additional-full.csv' inside the zip
with zip_file.open('bank-additional/bank-additional-full.csv') as f:
    # Read the CSV into a pandas DataFrame, specifying the separator
    df = pd.read_csv(f, sep=';')

print("Dataset loaded successfully!")
```
